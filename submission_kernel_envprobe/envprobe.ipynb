```
import glob, os
print('=== ALL wheels in /kaggle/input (version is in the filename) ===')
for w in glob.glob('/kaggle/input/**/*.whl', recursive=True): print('WHEEL', os.path.basename(w))
print('=== the installer __init__.py (what it pip-installs) ===')
for f in glob.glob('/kaggle/input/**/kaggle_evaluation/**/__init__.py', recursive=True) + glob.glob('/kaggle/input/**/kaggle_evaluation/__init__.py', recursive=True):
    t=open(f).read()
    if 'pip' in t and 'install' in t:
        print('--- FILE', f, '---')
        # print the install-related lines with context
        lines=t.splitlines()
        for i,l in enumerate(lines):
            if 'install' in l or 'whl' in l.lower() or 'glob' in l or 'wheel' in l.lower() or 'llama' in l.lower() or 'check_call' in l:
                print(f'  {i}: '+l.strip()[:120])
print('=== requirements/constraints files ===')
for r in glob.glob('/kaggle/input/**/requirement*.txt', recursive=True)+glob.glob('/kaggle/input/**/constraint*.txt', recursive=True):
    print('REQ', r); print(open(r).read()[:500])
print('WHEELPROBE DONE')
```
