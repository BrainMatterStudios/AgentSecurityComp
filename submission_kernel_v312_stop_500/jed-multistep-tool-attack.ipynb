## Run the gateway
Starts the inference server. During the competition rerun the gateway drives it against gpt-oss and gemma and writes `submission.csv`.

```python
import os
from pathlib import Path
sub = Path('/kaggle/working/submission.csv')
try:
    import kaggle_evaluation.jed_attack_134815.jed_attack_inference_server as srv
    srv.JEDAttackInferenceServer().serve()
except Exception as exc:
    if os.getenv('KAGGLE_IS_COMPETITION_RERUN') is not None:
        raise
    print('skipped (non-rerun):', repr(exc))
# Ensure submission.csv exists for the submit handshake; the gateway overwrites
# it with real per-model/per-guardrail scores during the competition rerun.
if not sub.exists():
    sub.write_text('Id,Score\n'
                   'gpt_oss_public,0.0\ngpt_oss_private,0.0\n'
                   'gemma_public,0.0\ngemma_private,0.0\n')
    print('wrote fallback submission.csv')
```
